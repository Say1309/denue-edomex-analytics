```python
import pandas as pd

# 1. Cargar el Parquet actual
df = pd.read_parquet("data/denue_edomex_para_app.parquet")

# 2. Función robusta para reparar el encoding (mojibake)
def reparar_texto(texto):
    if not isinstance(texto, str):
        return texto
    try:
        return texto.encode('latin-1').decode('utf-8')
    except (UnicodeEncodeError, UnicodeDecodeError):
        return texto

# 3. Aplicar limpieza a la columna de giros y estandarizar formato
if "nombre_act" in df.columns:
    df["nombre_act"] = df["nombre_act"].apply(reparar_texto).str.strip().str.title()
    
    # Ajustes finos de acentos comunes del SCIAN
    df["nombre_act"] = df["nombre_act"].str.replace("Administracion", "Administración", case=False)
    df["nombre_act"] = df["nombre_act"].str.replace("Publica", "Pública", case=False)
    df["nombre_act"] = df["nombre_act"].str.replace("Pensión", "Pensión", case=False)

# 4. Sobrescribir el Parquet con los datos ya limpios de una vez por todas
df.to_parquet("data/denue_edomex_para_app.parquet", index=False)

print("--- ¡PARQUET LIMPIADO Y ACTUALIZADO DEFINITIVAMENTE! ---")
```
